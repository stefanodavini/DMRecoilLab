# Spin-independent sensitivity for multi-ton argon and xenon detectors

This notebook illustrates the computation of projected
spin-independent WIMP sensitivities using the DMRecoilLab
sensitivity framework.

The calculation combines:
- elastic or inelastic scattering kinematics,
- spin-independent interactions,
- detector exposure and threshold effects,
- Poisson counting statistics.

The resulting curves are shown in the conventional
$(m_\chi,\sigma_p)$ plane.

## Elastic, endothermic and exothermic scattering

We compare three scenarios:

- elastic scattering ($\delta = 0$),
- endothermic scattering ($\delta = +100$ keV),
- exothermic scattering ($\delta = -100$ keV).

The comparison is shown for representative multi-ton
argon and xenon detectors.

Endothermic scattering suppresses low-velocity WIMPs and
typically weakens sensitivity, especially at low masses.

Exothermic scattering releases additional recoil energy
and can improve sensitivity in regions where detector
thresholds are important.

```python
import numpy as np
import matplotlib.pyplot as plt

from src.plotting.plots_upper_limits import plot_mass_sigma_limit_si
from src.detectors.reference_detectors import LZ_10YEAR, DS20K_10YEAR
```

```python
masses = np.geomspace(1e10,    # 10 GeV
                      1e12,    # 1000 GeV
                      50)
```

```python
delta_endo = 100e3  # 100 keV, endothermic
delta_exo = -100e3  # exothermic
```

```python
fig, ax = plot_mass_sigma_limit_si(masses=masses, detector=LZ_10YEAR,
                                   label="Multi-ton Xe, elastic",
                                   color="tab:blue",
                                   linestyle="-",
                                   linewidth=2.0)
fig, ax = plot_mass_sigma_limit_si(masses=masses, detector=LZ_10YEAR,
                                   delta=delta_endo,
                                   label=r"Multi-ton Xe, endothermic $\delta$=100 keV",
                                   color="tab:blue",
                                   linewidth=2.0,
                                   linestyle="--",
                                   ax=ax)
fig, ax = plot_mass_sigma_limit_si(masses=masses, detector=LZ_10YEAR,
                                   delta=delta_exo,
                                   label=r"Multi-ton Xe, exothermic $\delta$=-100 keV",
                                   color="tab:blue",
                                   linewidth=2.0,
                                   linestyle=":",
                                   ax=ax)
fig, ax = plot_mass_sigma_limit_si(masses=masses, detector=DS20K_10YEAR,
                                   label="Multi-ton Ar, elastic",
                                   color="tab:orange",
                                   linewidth=2.0,
                                   linestyle="-",
                                   ax=ax)
fig, ax = plot_mass_sigma_limit_si(masses=masses, detector=DS20K_10YEAR,
                                   delta=delta_endo,
                                   label=r"Multi-ton Ar, endothermic $\delta$=100 keV",
                                   color="tab:orange",
                                   linewidth=2.0,
                                   linestyle="--",
                                   ax=ax)
fig, ax = plot_mass_sigma_limit_si(masses=masses, detector=DS20K_10YEAR,
                                   delta=delta_exo,
                                   label=r"Multi-ton Ar, exothermic $\delta$=-100 keV",
                                   color="tab:orange",
                                   linewidth=2.0,
                                   linestyle=":",
                                   ax=ax)

ax.legend(title="Detector and interaction type")
```
