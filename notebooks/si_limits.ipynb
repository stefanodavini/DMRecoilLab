#SI Upper Limit Sensitivity

This notebook computes projected upper limits on the
WIMP-proton spin-independent cross section using the
DMRecoilLab sensitivity framework.

The calculation combines:
- elastic or inelastic scattering kinematics,
- spin-independent interactions,
- ideal detector response,
- Poisson counting statistics for zero counts.

The resulting curves are shown in the
$(m_\chi,\sigma_p)$ plane.

## Xenon vs Argon sensitivity

Compare projected sensitivities for argon and xenon
targets assuming either elastic scattering,
or inelastic with mass splitting
$$
\delta = \pm 100\ \mathrm{keV}.
$$

```python
import numpy as np
import matplotlib.pyplot as plt

from src.plotting.plots_upper_limits import plot_mass_sigma_limit_si
from src.detectors.reference_detectors import LZ_10YEAR, DS20K_10YEAR
```

```python
masses = np.geomspace(1e10,    # 10 GeV
                      1e12,    # 1000 GeV
                      50)
```

```python
delta_endo = 100e3  # 100 keV, endothermic
delta_exo = -100e3  # exothermic

fig, ax = plot_mass_sigma_limit_si(masses=masses, detector=LZ_10YEAR,
                                   label="Multi-ton Xe, elastic",
                                   color="tab:green",
                                   linewidth=2.5)
fig, ax = plot_mass_sigma_limit_si(masses=masses, detector=LZ_10YEAR,
                                   delta=delta_endo,
                                   label=r"Multi-ton Xe, endothermic $\delta$=100 keV",
                                   color="tab:green",
                                   linewidth=2.5,
                                   linestyle="--",
                                   ax=ax)
fig, ax = plot_mass_sigma_limit_si(masses=masses, detector=LZ_10YEAR,
                                   delta=delta_exo,
                                   label=r"Multi-ton Xe, exothermic $\delta$=-100 keV",
                                   color="tab:green",
                                   linewidth=2.5,
                                   linestyle=":",
                                   ax=ax)
fig, ax = plot_mass_sigma_limit_si(masses=masses, detector=DS20K_10YEAR,
                                   label="Multi-ton Ar, elastic",
                                   color="tab:orange",
                                   linewidth=2.5,
                                   ax=ax)
fig, ax = plot_mass_sigma_limit_si(masses=masses, detector=DS20K_10YEAR,
                                   delta=delta_endo,
                                   label=r"Multi-ton Ar, endothermic $\delta$=100 keV",
                                   color="tab:orange",
                                   linewidth=2.5,
                                   linestyle="--",
                                   ax=ax)
fig, ax = plot_mass_sigma_limit_si(masses=masses, detector=DS20K_10YEAR,
                                   delta=delta_exo,
                                   label=r"Multi-ton Ar, exothermic $\delta$=-100 keV",
                                   color="tab:orange",
                                   linewidth=2.5,
                                   linestyle=":",
                                   ax=ax)
```
