# Nuclear form factors

At sufficiently large momentum transfer the nucleus can no longer be treated as a point-like object.

Nuclear form factors suppress coherent scattering and become increasingly important at high recoil energies and for heavy target nuclei.

This notebook compares the Helm form factor for several commonly used detector materials.

## Momentum-transfer dependence

The Helm form factor approaches unity at small momentum transfer.

At larger momentum transfer the nuclear structure becomes important, leading to suppression and characteristic diffraction minima.

```python
from src.nuclei.form_factors import HelmFormFactor
from src.nuclei.isotopes import Xe131, Ar40, Na23
from src.kinematics import ElasticKinematics

from src.plotting.plots_form_factors import plot_response_vs_ER, plot_response

import numpy as np
import matplotlib.pyplot as plt

targets = [Xe131, Ar40, Na23]
helms = []

for nucleus in targets:
    kin = ElasticKinematics(mX=1e12, mT=nucleus.mass)
    helms.append(HelmFormFactor(nucleus=nucleus))

q = np.linspace(0, 5e8, 1000)

fig, ax = plot_response(form_factor=helms[0], q=q, label=r"$^{131}$Xe")
fig, ax = plot_response(form_factor=helms[1], q=q, label=r"$^{40}$Ar", ax=ax)
fig, ax = plot_response(form_factor=helms[2], q=q, label=r"$^{23}$Na", ax=ax)

plt.show()
```

## Recoil-energy dependence

The same form-factor suppression can be expressed as a function of recoil energy using

$$
q = \sqrt{2 m_T E_R}
$$

Heavy nuclei experience stronger suppression at a fixed recoil energy because the corresponding momentum transfer is larger.

```python
ER = np.linspace(0, 3e5, 1000)

targets = [Xe131, Ar40, Na23]
helms = []
kins = []

for nucleus in targets:
    kins.append(ElasticKinematics(mX=1e12, mT=nucleus.mass))
    helms.append(HelmFormFactor(nucleus=nucleus))

fig, ax = plot_response_vs_ER(form_factor=helms[0], kinematics=kins[0],
                              ER=ER, label=r"$^{131}$Xe")
fig, ax = plot_response_vs_ER(form_factor=helms[1], kinematics=kins[1],
                              ER=ER, label=r"$^{40}$Ar", ax=ax)
fig, ax = plot_response_vs_ER(form_factor=helms[2], kinematics=kins[2],
                              ER=ER, label=r"$^{23}$Na", ax=ax)


plt.show()
```
