# Halo models

The recoil spectrum observed by a dark-matter detector depends not only on the particle physics interaction, but also on the velocity distribution of dark matter in the Solar neighbourhood.

This notebook illustrates the Standard Halo Model (SHM) and the effect of boosting the Galactic velocity distribution into the detector frame.

The plots below show speed distributions normalized such that

$$
\int_0^\infty g(v)\,dv = 1.
$$

All velocities are displayed in km/s.

## Earth's motion through the halo

Direct-detection experiments do not observe the Galactic-frame velocity distribution.

Instead, the detector moves through the Milky Way halo with a speed of approximately 230 km/s.

The boost shifts the distribution toward larger observed speeds and enhances the population of high-velocity WIMPs that are most relevant for producing energetic nuclear recoils.

```python
from src.plotting.plots_halo import plot_boosted_speed_distribution

fig, ax = plot_boosted_speed_distribution(shm, vboost=232)

ax.set_yscale("log")
plt.show()
```
