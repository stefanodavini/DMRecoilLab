## Rate-only interpretation

For each dark matter model, the interaction cross section is normalized
such that the expected number of LZ events equals the observed count.
The corresponding prediction for DS is then computed.
This procedure uses only the total event count and ignores recoil-energy information.

```python
import numpy as np
import matplotlib.pyplot as plt

from src.analysis.experiment_matching import matched_counts_grid_si
from src.plotting.plots_experiment_matching import plot_matched_counts

from src.detectors.reference_detectors import LZ_2024, DS20K_10YEAR
```

## Single benchmark

SI interaction model with exothermic kinematics.

```python
grid = matched_counts_grid_si(
    wimp_masses=[100e9],        # masses in eV
    deltas=[-1e5],              # inelastic delta splitting 
    counts_a=1,                 # counts observed by detector A
    detector_a=LZ_2024,         # detector A
    detector_b=DS20K_10YEAR,    # detector B
)

counts_b = grid.counts_b
print("DS20k will observe ", counts_b)
```

## Mass scan, fixed $\delta$

```python
masses = np.geomspace(1e10, 1e12, 50)

grid = matched_counts_grid_si(
    wimp_masses=masses,
    deltas=[-1e5],
    counts_a=10,
    detector_a=LZ_2024,
    detector_b=DS20K_10YEAR,
)
```

```python
plt.semilogx(
    grid.masses,
    grid.counts_b[:, 0]
)
plt.xlabel("WIMP mass (eV)")
plt.ylabel("Counts in DS-20k")
```

# Mass-$\delta$ scan

```python
masses = np.geomspace(1e10, 1e12, 20)

deltas = np.linspace(-200e3, 300e3, 50)

grid = matched_counts_grid_si(wimp_masses=masses,
                              deltas=deltas,
                              counts_a=1,
                              detector_a=LZ_2024,
                              detector_b=DS20K_10YEAR)
```

Regions with large matched counts correspond
to parameter combinations for which DS20k
is more sensitive than LZ.

Regions with small matched counts indicate
parameter combinations for which the signal
is suppressed in the DS20k detector.

```python
plot_matched_counts(grid=grid)
```
