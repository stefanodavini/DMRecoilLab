# Scattering kinematics

Nuclear recoil energies are constrained by energy and momentum conservation.

For a given WIMP speed, only a limited range of recoil energies is physically accessible.

This notebook illustrates the kinematic boundaries of elastic and inelastic WIMP-nucleus scattering.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
```

## Allowed recoil-energy region

The region between the minimum and maximum recoil energy defines the kinematically accessible phase space.

Elastic and inelastic interactions produce markedly different recoil-energy boundaries, particularly at low WIMP speeds.


```python
import matplotlib.pyplot as plt
import numpy as np

from src.plotting.plots_kinematics import plot_ER_bounds, plot_elastic_vs_inelastic_vmin
from src.utils.factory import create_kinematics
from src.nuclei.isotopes import Ar40
from src.interactions.particles import WIMP

wimp = WIMP(mass=1e12, spin=0)

elastic = create_kinematics(wimp=wimp, nucleus=Ar40)
vgrid = np.linspace(0, 1e-3, 1000)
fig, ax = plot_ER_bounds(kin=elastic, vgrid=vgrid)

plt.show()
```

```python

inelastic = create_kinematics(wimp=wimp, nucleus=Ar40, delta=2e5)
vthr = inelastic.threshold_speed
vgrid = np.linspace(vthr + 1e-5, vthr + 1e-3, 500)
fig, ax = plot_ER_bounds(kin=inelastic, vgrid=vgrid)

plt.show()
```

## Minimum velocity

The quantity
$$
v_{\min}(E_R)
$$

plays a central role in direct detection.

It determines the minimum incoming WIMP speed required to produce a recoil of energy \(E_R\).

For endothermic inelastic scattering ($\delta > 0$), an additional energy cost must be paid to excite the dark sector, leading to a significant increase in $v_{\min}$.

```python
ERgrid = np.linspace(1e4, 3e5, 100)
fig, ax = plot_elastic_vs_inelastic_vmin(elastic=elastic, inelastic=inelastic, ER=ERgrid)

plt.show()
```
